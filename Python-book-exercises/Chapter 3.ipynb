```python
import matplotlib.pyplot as plt
import numpy as np
```

# Q5

```python
N = 200
k = 30

d1 = np.random.normal(loc=2, scale=1, size=N)
d2 = np.exp(np.random.normal(size=N))

d = np.hstack((d1, d2))
bins = np.linspace(np.min(d), np.max(d), k + 1)

y1, x1 = np.histogram(d1, bins=bins)
y2, x2 = np.histogram(d2, bins=bins)
xx = (x1[:-1] + x1[1:]) / 2

_, axs = plt.subplots(1, 2, figsize=(10, 3))

# draw the histograms using matplotlib
axs[0].hist(d1, bins=bins, label='Gaussian')
axs[0].hist(d2, bins=bins, label='Exponential')
axs[0].set_title(r'A) Histograms using bars')

# now draw lines
axs[1].plot(xx, y1, linewidth=3, label='Gaussian')
axs[1].plot(xx, y2, linewidth=3, label='Exponential')
axs[1].set_title(r'B) Histograms using lines')

for a in axs:
    a.set_xlabel('Data value')
    a.set_ylabel('Count')
    a.legend()

plt.tight_layout()
plt.show()
```
